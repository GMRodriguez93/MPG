# Ejercicio 06

Una muestra aleatoria de 400 domicilios de una ciudad mostró que 25% son casas alquiladas. ¿Cuál será el número de casas alquiladas de esa ciudad, a un nivel de significación del 2%, suponiendo que la ciudad tiene 20 000 casas?

Se estima el número de casas alquiladas mediante un intervalo de confianza del 98% para una proporción. Se supone que la muestra fue seleccionada sin reemplazo; por eso se incluye la corrección por población finita.


## Paso a paso

1. Proporción muestral de casas alquiladas: p̂ = 0.25; por lo tanto, q̂ = 1 - p̂ = 0.75.
2. Tamaño de muestra: n = 400; tamaño de la población: N = 20 000.
3. Nivel de significación: α = 0.02. El nivel de confianza es 1 - α = 0.98 (98%).
4. Valor crítico para un intervalo bilateral: z₁₋α/₂ = z₀.₉₉ ≈ 2.3263.
5. Error estándar con corrección por población finita: √[(p̂ q̂ / n) ((N - n) / (N - 1))].
6. Margen de error para la proporción: E = z₁₋α/₂ × error estándar.
7. Intervalo para la proporción de casas alquiladas: p̂ ± E.
8. Para estimar cantidades de casas, multiplicamos los extremos del intervalo por N. La estimación puntual es 0.25 × 20 000 = 5 000 casas.

La aproximación normal es adecuada porque la muestra contiene aproximadamente 100 domicilios alquilados y 300 no alquilados. Los límites de cantidad se redondean hacia afuera para expresar el intervalo en casas enteras.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

n = 400
p_hat = 0.25
q_hat = 1 - p_hat
N = 20_000
alpha = 0.02
confianza = 1 - alpha

z_critico = norm.ppf(1 - alpha / 2)
correccion_poblacion_finita = np.sqrt((N - n) / (N - 1))
error_estandar = np.sqrt((p_hat * q_hat / n) * ((N - n) / (N - 1)))
margen_error = z_critico * error_estandar
limite_inferior_p = p_hat - margen_error
limite_superior_p = p_hat + margen_error

estimacion_casas = p_hat * N
limite_inferior_casas = limite_inferior_p * N
limite_superior_casas = limite_superior_p * N

print(f'Nivel de confianza: {confianza:.0%}')
print(f'Valor crítico z: {z_critico:.4f}')
print(f'Corrección por población finita: {correccion_poblacion_finita:.4f}')
print(f'Error estándar de la proporción: {error_estandar:.5f}')
print(f'Margen de error de la proporción: {margen_error:.5f}')
print(f'IC del 98% para la proporción: ({limite_inferior_p:.4f}, {limite_superior_p:.4f})')
print(f'Estimación puntual: {estimacion_casas:,.0f} casas alquiladas')
print(f'IC del 98% para el número de casas: ({limite_inferior_casas:,.2f}, {limite_superior_casas:,.2f})')
print(f'En casas enteras: entre {np.floor(limite_inferior_casas):,.0f} y {np.ceil(limite_superior_casas):,.0f} casas')

desvio_casas = N * error_estandar
x = np.linspace(estimacion_casas - 4 * desvio_casas, estimacion_casas + 4 * desvio_casas, 2000)
y = norm.pdf(x, loc=estimacion_casas, scale=desvio_casas)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, y, color='navy', lw=2, label='Aproximación normal de la estimación')
ax.fill_between(
    x, y,
    where=(x >= limite_inferior_casas) & (x <= limite_superior_casas),
    color='skyblue', alpha=0.7, label='Región central del 98%'
)
ax.axvline(limite_inferior_casas, color='red', linestyle='--', label=f'Límite inferior = {limite_inferior_casas:,.0f}')
ax.axvline(limite_superior_casas, color='red', linestyle='--', label=f'Límite superior = {limite_superior_casas:,.0f}')
ax.axvline(estimacion_casas, color='green', linestyle='-.', label=f'Estimación puntual = {estimacion_casas:,.0f}')
ax.set_title('Estimación del número de casas alquiladas (IC del 98%)')
ax.set_xlabel('Número de casas alquiladas en la ciudad')
ax.set_ylabel('Densidad aproximada')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Interpretación y respuesta

La mejor estimación puntual es **5 000 casas alquiladas**. Con un nivel de confianza del 98% y aplicando la corrección por población finita, se estima que hay entre **aproximadamente 4 002 y 5 998 casas alquiladas** en la ciudad.

Este intervalo expresa la incertidumbre de la estimación; no permite conocer el número exacto de casas alquiladas sin censar todos los domicilios.
