# Ejercicio 03

Siendo X = 500, S = 16 y n = 25. Determinar el intervalo de confianza a un nivel de seguridad del 98%.

En este ejercicio se interpreta S = 16 como la desviación estándar poblacional conocida (σ). Por lo tanto, la varianza poblacional es σ² = 16² = 256 y corresponde utilizar la distribución normal estándar Z.


## Paso a paso

1. Media muestral: x̄ = 500
2. Desviación estándar poblacional conocida: σ = 16
3. Varianza poblacional: σ² = 16² = 256
4. Tamaño muestral: n = 25
5. Nivel de confianza: 98% => α = 1 - 0.98 = 0.02; α/2 = 0.01
6. Valor crítico: z_{1-α/2} = z_{0.99} ≈ 2.3263
7. Error estándar: σ / √n = 16 / √25 = 3.2
8. Margen de error: E = z * σ/√n = 2.3263 * 3.2 ≈ 7.44
9. Intervalo de confianza: x̄ ± E = 500 ± 7.44 = (492.56, 507.44)


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x_bar = 500
S = 16
n = 25
confianza = 0.98
alpha = 1 - confianza
varianza_poblacional = S**2
sigma = np.sqrt(varianza_poblacional)
z_crit = norm.ppf(1 - alpha / 2)
error_estandar = sigma / np.sqrt(n)
error = z_crit * error_estandar
li = x_bar - error
ls = x_bar + error

print(f'Nivel de confianza: {confianza * 100:.0f}%')
print(f'Varianza poblacional: {varianza_poblacional:.0f}')
print(f'Valor crítico z: {z_crit:.4f}')
print(f'Error estándar: {error_estandar:.4f}')
print(f'Error de estimación: {error:.4f}')
print(f'IC = ({li:.2f}, {ls:.2f})')

x = np.linspace(x_bar - 4 * error_estandar, x_bar + 4 * error_estandar, 2000)
y = norm.pdf(x, loc=x_bar, scale=error_estandar)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, y, color='navy', lw=2, label='Distribución de la media muestral')
ax.fill_between(x, y, where=(x >= li) & (x <= ls), color='skyblue', alpha=0.7, label='Región central del 98%')
ax.axvline(li, color='red', linestyle='--', label=f'LI = {li:.2f}')
ax.axvline(ls, color='red', linestyle='--', label=f'LS = {ls:.2f}')
ax.axvline(x_bar, color='green', linestyle='-.', label=f'Media = {x_bar:.0f}')
ax.set_title('Intervalo de confianza del 98% usando Z')
ax.set_xlabel('Media muestral')
ax.set_ylabel('Densidad')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Resultado

Interpretando S = 16 como la desviación estándar poblacional conocida, con un nivel de confianza del 98%, el intervalo de confianza para la media poblacional es:

IC = (492.56, 507.44)
