# Z test para la proporción

Prueba de una muestra para verificar si la proporción poblacional es igual a un valor especificado.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Datos
p0 = 0.5
x_exitos = 12
n = 20
alpha = 0.05
tipo_prueba = 'dos_colas'  # 'dos_colas', 'mayor', 'menor'

# Hipotesis
# H0: p = p0
# H1: p != p0 (o mayor/menor segun tipo_prueba)

p_hat = x_exitos / n
z = (p_hat - p0) / np.sqrt((p0 * (1 - p0)) / n)
p_valor = 2 * stats.norm.sf(abs(z)) if tipo_prueba == 'dos_colas' else stats.norm.sf(z) if tipo_prueba == 'mayor' else stats.norm.cdf(z)
z_critico = stats.norm.ppf(1 - alpha/2) if tipo_prueba == 'dos_colas' else stats.norm.ppf(1 - alpha) if tipo_prueba == 'mayor' else stats.norm.ppf(alpha)

print(f'Proporcion hipotetica bajo H0: {p0}')
print(f'Proporcion muestral: {p_hat}')
print(f'Tamanio muestral: {n}')
print(f'Z calculado: {z:.4f}')
print(f'Valor p: {p_valor:.4f}')
print(f'Valor critico: {z_critico:.4f}')

if p_valor < alpha:
    print('Se rechaza H0')
else:
    print('No se rechaza H0')

# Grafica
x = np.linspace(-4, 4, 1000)
y = stats.norm.pdf(x)
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, y, color='navy', lw=2, label='Distribucion normal')

if tipo_prueba == 'dos_colas':
    ax.fill_between(x, y, where=(x <= -z_critico) | (x >= z_critico), color='red', alpha=0.35, label='Region critica')
    ax.axvline(-z_critico, color='red', linestyle='--')
    ax.axvline(z_critico, color='red', linestyle='--')
elif tipo_prueba == 'mayor':
    ax.fill_between(x, y, where=x >= z_critico, color='red', alpha=0.35, label='Region critica')
    ax.axvline(z_critico, color='red', linestyle='--')
else:
    ax.fill_between(x, y, where=x <= z_critico, color='red', alpha=0.35, label='Region critica')
    ax.axvline(z_critico, color='red', linestyle='--')

ax.axvline(z, color='darkorange', lw=2.5, label=f'Z calculado = {z:.3f}')
ax.set_title('Prueba z para la proporción')
ax.set_xlabel('Valor del estadistico z')
ax.set_ylabel('Densidad')
ax.grid(alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()
```
