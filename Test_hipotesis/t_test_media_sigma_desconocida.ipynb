# T test para la media (sigma desconocida)

Prueba de una muestra para verificar si la media poblacional es igual a un valor especificado cuando la desviacion poblacional no es conocida.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Datos
mu0 = 115
Media_muestral = 118 
s = 20                      #Medida de dispersion (desviacion estandar muestral)
n = 20
alpha = 0.05                #Nivel de significacion
tipo_prueba = 'dos_colas'  # 'dos_colas', 'mayor', 'menor'

# Hipotesis
# H0: mu = mu0
# H1: mu != mu0 (o mayor/menor segun tipo_prueba)

gl = n - 1
t = (Media_muestral - mu0) / (s / np.sqrt(n))
p_valor = 2 * stats.t.sf(abs(t), df=gl) if tipo_prueba == 'dos_colas' else stats.t.sf(t, df=gl) if tipo_prueba == 'mayor' else stats.t.cdf(t, df=gl)
t_critico = stats.t.ppf(1 - alpha/2, df=gl) if tipo_prueba == 'dos_colas' else stats.t.ppf(1 - alpha, df=gl) if tipo_prueba == 'mayor' else stats.t.ppf(alpha, df=gl)

print(f'Media hipotetica bajo H0: {mu0}')
print(f'Media muestral: {Media_muestral}')
print(f'Desviacion estandar muestral: {s}')
print(f'Tamanio muestral: {n}')
print(f'Grados de libertad: {gl}')
print(f'T calculado: {t:.4f}')
print(f'Valor p: {p_valor:.4f}')
print(f'Valor critico: {t_critico:.4f}')

if p_valor < alpha:
    print('Se rechaza H0')
else:
    print('No se rechaza H0')

# Grafica
x = np.linspace(stats.t.ppf(0.001, df=gl), stats.t.ppf(0.999, df=gl), 1000)
y = stats.t.pdf(x, df=gl)
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, y, color='navy', lw=2, label='Distribucion t de Student')

if tipo_prueba == 'dos_colas':
    ax.fill_between(x, y, where=(x <= -t_critico) | (x >= t_critico), color='red', alpha=0.35, label='Region critica')
    ax.axvline(-t_critico, color='red', linestyle='--')
    ax.axvline(t_critico, color='red', linestyle='--')
elif tipo_prueba == 'mayor':
    ax.fill_between(x, y, where=x >= t_critico, color='red', alpha=0.35, label='Region critica')
    ax.axvline(t_critico, color='red', linestyle='--')
else:
    ax.fill_between(x, y, where=x <= t_critico, color='red', alpha=0.35, label='Region critica')
    ax.axvline(t_critico, color='red', linestyle='--')

ax.axvline(t, color='darkorange', lw=2.5, label=f'T calculado = {t:.3f}')
ax.set_title('Prueba t para la media con sigma desconocida')
ax.set_xlabel('Valor del estadistico t')
ax.set_ylabel('Densidad')
ax.grid(alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()
```
