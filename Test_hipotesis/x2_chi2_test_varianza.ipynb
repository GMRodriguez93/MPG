# Chi-cuadrado para la varianza

Prueba de una muestra para verificar si la varianza poblacional es igual a un valor hipotetico.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Datos
varianza_h0 = 144
s = 20
n = 20
alpha = 0.05
tipo_prueba = 'dos_colas'  # 'dos_colas', 'mayor', 'menor'

# Hipotesis
# H0: sigma^2 = varianza_h0
# H1: sigma^2 != varianza_h0 (o mayor/menor segun tipo_prueba)

gl = n - 1
varianza_muestral = s ** 2
chi2 = gl * varianza_muestral / varianza_h0
p_valor = 2 * stats.chi2.sf(abs(chi2), df=gl) if tipo_prueba == 'dos_colas' else stats.chi2.sf(chi2, df=gl) if tipo_prueba == 'mayor' else stats.chi2.cdf(chi2, df=gl)

print(f'Varianza hipotetica bajo H0: {varianza_h0}')
print(f'Varianza muestral: {varianza_muestral}')
print(f'Grados de libertad: {gl}')
print(f'Chi-cuadrado calculado: {chi2:.4f}')
print(f'Valor p: {p_valor:.4f}')

if p_valor < alpha:
    print('Se rechaza H0')
else:
    print('No se rechaza H0')

# Grafica
x = np.linspace(stats.chi2.ppf(0.001, df=gl), stats.chi2.ppf(0.999, df=gl), 1000)
y = stats.chi2.pdf(x, df=gl)
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, y, color='navy', lw=2, label='Distribucion chi-cuadrado')

if tipo_prueba == 'dos_colas':
    critico_izq = stats.chi2.ppf(alpha/2, df=gl)
    critico_der = stats.chi2.ppf(1 - alpha/2, df=gl)
    ax.fill_between(x, y, where=(x <= critico_izq) | (x >= critico_der), color='red', alpha=0.35, label='Region critica')
    ax.axvline(critico_izq, color='red', linestyle='--')
    ax.axvline(critico_der, color='red', linestyle='--')
elif tipo_prueba == 'mayor':
    critico = stats.chi2.ppf(1 - alpha, df=gl)
    ax.fill_between(x, y, where=x >= critico, color='red', alpha=0.35, label='Region critica')
    ax.axvline(critico, color='red', linestyle='--')
else:
    critico = stats.chi2.ppf(alpha, df=gl)
    ax.fill_between(x, y, where=x <= critico, color='red', alpha=0.35, label='Region critica')
    ax.axvline(critico, color='red', linestyle='--')

ax.axvline(chi2, color='darkorange', lw=2.5, label=f'Chi-cuadrado calculado = {chi2:.3f}')
ax.set_title('Prueba chi-cuadrado para la varianza')
ax.set_xlabel('Valor del estadistico chi-cuadrado')
ax.set_ylabel('Densidad')
ax.grid(alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()
```
